```python
sns.set(style="whitegrid")

plt.figure(figsize=(12,6))

sample = data1.head(10)

sns.boxplot(data=sample, y='Stress_Level')

plt.title('Box Plot of Stress Level')
plt.ylabel('Stress Level')

plt.show()
```
